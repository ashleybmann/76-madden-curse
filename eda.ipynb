## Personal EDA

There are 2 CSVs that I've created with Claude. The first is events. The second is page views.

Events:
- An "event" is one player plus one announcement. Players who were named more than once get a row for each time. Aaron Donald, for example, has several All-Pro rows, so 323 events cover 197 players.
- Key columns:
    - event_type: madden_cover or ap_all_pro_1st. This is the treatment vs. comparison split.
    - event_date: the announcement day, which is day 0.
    - window_start / window_end: the 90 days before through 180 days after.
    - wiki_title: the Wikipedia page the views came from.
    - include: whether the event is kept or excluded (e.g., Madden 23).
    - overlaps_window / overlap_days: flags players whose windows from two events overlap.
- event_id is the unique label that links the two files.

Page views: 
- pageviews_daily.csv: one row per event per day (85,244 rows)
- There are just four columns: event_id, date, rel_day, views.
- rel_day is the number of days relative to the announcement, from −90 to +180. This lets you line up every event on the same timeline, whatever the calendar date.
- Most events have 271 rows. That's −90 through +180 counting day 0, so the window is 271 days, not 270.
- Only 315 of the 323 events have views. The other 8 are mostly the excluded rows, and a few windows are shorter than 271 days.

```python
# Re-read with the date columns parsed as real dates (otherwise they're plain text)
events = pd.read_csv(
    '/Users/AMann/Desktop/Coding/Afternoon Projects/04_76_Madden Curse/data/events.csv',
    parse_dates=['event_date', 'window_start', 'window_end'],
)

# Overall earliest/latest for each date column
display(events[['event_date', 'window_start', 'window_end']].agg(['min', 'max']))

# Announcement date range by event type
events.groupby('event_type')['event_date'].agg(['min', 'max', 'nunique', 'count'])
```

```python
pageviews = pd.read_csv(
    '/Users/AMann/Desktop/Coding/Afternoon Projects/04_76_Madden Curse/data/pageviews_daily.csv',
    parse_dates=['date'],
)

# Overall calendar range and relative-day range
display(pageviews[['date', 'rel_day']].agg(['min', 'max']))

# Per-event coverage: first/last day and how many days of data each event has
per_event = pageviews.groupby('event_id').agg(
    first_date=('date', 'min'),
    last_date=('date', 'max'),
    first_rel=('rel_day', 'min'),
    last_rel=('rel_day', 'max'),
    n_days=('date', 'size'),
)
display(per_event['n_days'].value_counts().sort_index())

# Events missing days (a full window is 271 days: -90 through +180)
per_event[per_event['n_days'] < 271]
```
